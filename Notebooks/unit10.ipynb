## Insertion sort

Insertion sort has a linear run time in the best case and a quadratic run time in the worst case.

```python
import random
import timeit
import matplotlib.pyplot as plt


def insertion_sort(l):
    if len(l) < 2:
        return
    
    for i in range(1, len(l)):
        current_value = l[i]
        j = i - 1
        while j >= 0 and current_value < l[j]:
            l[j + 1] = l[j]
            j -= 1
        l[j + 1] = current_value
```

### Testing insertion sort

```python
def test():
    l = [4, 3, 1, 2, 6]
    insertion_sort(l)
    print(l)
    

test()
```

### Measuring time for insertion sort in the best and worst case

```python
# Benchmarking insertion sort
n_max = 1000
avg_sorted = []
avg_reversed = []

for n in range(100, n_max, 100):
    print(n)
    l = random.sample(range(1, 100000), n)
    l_sorted = l.copy()
    l_sorted.sort() # Best case: sorted already
    l_reversed = l.copy()
    l_reversed.sort(reverse=True) # Worst case: sorted in reverse order
    
    result_sorted = []
    result_reversed = []

    # Run each case 100 times and take the average
    for i in range(100):
        l1 = l_sorted.copy()
        result_sorted.append(timeit.timeit('insertion_sort(l1)', globals=globals(), number=1))
        l1 = l_reversed.copy()
        result_reversed.append(timeit.timeit('insertion_sort(l1)', globals=globals(), number=1))
   
    avg_sorted.append(1000 * (sum(result_sorted) / 100))
    avg_reversed.append(1000 * (sum(result_reversed) / 100))

plt.plot(range(100, n_max, 100), avg_sorted, label='Best case')
plt.plot(range(100, n_max, 100), avg_reversed, label='Worst case')
plt.title('Insertion sort run time')
plt.xlabel('Input size $n$')
plt.ylabel('Time in ms')
plt.legend()
```

### Binary Search ###

The binary search algorithm is a fast algorithm that checks the occurrence of an item in a sorted list.

```python
def binary_search(l, o):
    low = 0
    high = len(l) - 1
    while low <= high:
        mid = low + (high - low)//2
        if l[mid] == o:
            return True
        elif l[mid] < o:
            low = mid + 1
        else:
            high = mid - 1
    return False

l = [1, 3, 5, 7, 9]
print(binary_search(l, 5))
print(binary_search(l, 6))
```
