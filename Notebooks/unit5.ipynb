# Tuples (immutable, ordered)

Tuples are _ordered_ and _immutable_. Ordered means, that an element at list position i will always be at this position. Immutable means that we cannot change a tuple (in contrast to a list)

```python
fixed_numbers = (1, 2, 3)
print(fixed_numbers)
```

```python
fixed_numbers = (1, 2, 3)
print(f'Indexing is similar to lists: {fixed_numbers[1]}')
```

```python
fixed_numbers = (1, 2, 3)
print(f'Indexing is similar to lists, another example: {fixed_numbers[0:2]}')
```

```python
# Same function with len
fixed_numbers = (1, 2, 3)
print('Length ==', len(fixed_numbers))
```

```python
# BUT: tuples are immutable
fixed_numbers = (1, 2, 3)
fixed_numbers[0] = 13
print(fixed_numbers)
```

```python
# Associated values
person = ("Alice", 25)

# Fixed sequences
rgb = (255, 0, 0)

# When data stays constant
days = ("Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun")
```

Tuples have **less internal overhead** than lists. Therefore, they can be used to
store data that does not need to be changed, leading to performance improvements
in certain situations. Additionally, using tuples can help ensure data integrity
by **preventing accidental modifications**.

The below code uses the 'sys' module (more on modules later) to compare the memory size of a list and a tuple.

```python
import sys
a = [1, 2, 3] # a list
b = (1, 2, 3) # a tuple

print("List a:", sys.getsizeof(a), "bytes")  # e.g., 88 bytes
print("Tuple b:", sys.getsizeof(b), "bytes")  # e.g., 64 bytes
```

```python
# Continue "skips" the rest of the current iteration and continues with the next element
lectures = ['maths', 'english', 'computer science', 'geology', 'physics']

for lecture in lectures:
    if lecture == 'maths' or lecture == 'physics':
        continue # skip right to next item if looking at Maths or Physics
    print(f'I love {lecture}!')
```

### Flags, conditions, and while loops

```python
# "flags", option one
count = 0
is_not_ten = True
while is_not_ten:
    print(f"Current count: {count}")
    count += 1
    is_not_ten = (count != 10)
```

```python
# "flags", option two
count = 0
is_not_ten = True
while is_not_ten:
    print(f"Current count: {count}")
    count += 1

    if count == 10:
        is_not_ten = False
```

```python
# another option:
# Endless loop with break
count = 0
while True:
    print(f"Current count: {count}")
    count += 1

    if count == 10:
        break
```

# Putting it all together: basic linear regression with gradient descent

```python
n = int(input('Enter the number of data points: '))

data = []
count = 0
while count < n:
    x = float(input(f'Enter the x coordinate of the {count}-th point: '))
    y = float(input(f'Enter the y coordinate of the {count}-th point: '))
    data.append((x, y))
    count += 1

print(data)

tol = 1e-33
# try tol = 1e-3
learning_rate = 1e-2
a = 10
b = 5
error = 0
error_old = 1
count = 0

while abs(error_old - error) > tol:
    a_grad = 0
    b_grad = 0
    for point in data:
        a_grad += (a * point[0] + b - point[1]) * point[0]
        b_grad += a * point[0] + b - point[1]
    a -= learning_rate * a_grad  # learning_rate_new = 2 * learning_rate
    b -= learning_rate * b_grad

    error_old = error
    error = 0
    for point in data:
        error += ((a * point[0] + b) - point[1]) ** 2
    count += 1
    print(f'Error difference {count}: {abs(error_old-error)}')
    print(f'\ta={a}, b={b}')

print(f'Slope of the best fitting line is {a}.')
print(f'Intercept of the best fitting line is {b}.')
print(f'The error is {error}.')
print(f'The algorithm converged in {count} steps.')
```

```python
import matplotlib.pyplot as plt
x_vals = [point[0] for point in data]
y_vals = [point[1] for point in data]

plt.scatter(x_vals, y_vals, color='blue', label='Data points')
x_line = [min(x_vals), max(x_vals)]
y_line = [a * x + b for x in x_line]

plt.plot(x_line, y_line, color='red', label='Fitted line')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Linear Regression Fit')
plt.legend()

plt.show()
```
