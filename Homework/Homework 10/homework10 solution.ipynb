# Running times of loops

Write a small Python script to measure the running time of loops.  First, define a number $n$.  Before each loop define a counter variable and set it to zero.  In each step of the loop increment the counter by 1.   After the loop has finished, print the counter.

Implement the following loops:

1. **Loop A.**  
   Write a `while` loop that starts with the loop variable being equal to $n$ and runs until the loop variable is larger than 0.  Within the loop, integer divide the loop variable by two.

2. **Loop B.**  
   Write a single `for` loop from 0 to $n - 1$.

3. **Loop C.**  
   Write a nested loop with Loop B as the outer loop and Loop A as the inner loop.

4. **Loop D.**  
   Write a nested loop with Loop B as the outer loop and Loop B as the inner loop, i.e., a nested loop with two `for` loops.

5. **Loop E.**  
   Write a nested loop with Loop D as the outer loop and Loop A as the inner loop, i.e., Loop A is enclosed in two `for` loops.

6. **Loop F.**  
   Write a nested loop with three `for` loops.

---

### **Instructions**

In your code, write comments after printing the results and discuss the running times you have obtained.

```python
import time

# Choose a value of n
n = 1000

print(f"n = {n}\n")

############################
# 1) Loop A
############################
counter_A = 0
x = n

start = time.time()
while x > 0:
    counter_A += 1
    x //= 2
end = time.time()

print("Loop A counter:", counter_A)
print("Loop A time:   ", end - start)
# Loop A halves x each iteration → runs about log2(n) times. So: Θ(log n).
print()


############################
# 2) Loop B
############################
counter_B = 0

start = time.time()
for i in range(n):
    counter_B += 1
end = time.time()

print("Loop B counter:", counter_B)
print("Loop B time:   ", end - start)
# Loop B runs n times → Θ(n).
print()


############################
# 3) Loop C (B outside, A inside)
############################
counter_C = 0

start = time.time()
for i in range(n):          # outer: n times
    x = n
    while x > 0:            # inner: log n times
        counter_C += 1
        x //= 2
end = time.time()

print("Loop C counter:", counter_C)
print("Loop C time:   ", end - start)
# Outer loop: n iterations
# Inner loop: log n iterations
# Total: Θ(n log n).
print()


############################
# 4) Loop D (two for-loops)
############################
counter_D = 0

start = time.time()
for i in range(n):          # outer: n
    for j in range(n):      # inner: n
        counter_D += 1
end = time.time()

print("Loop D counter:", counter_D)
print("Loop D time:   ", end - start)
# Nested n × n loops → Θ(n^2).
print()


############################
# 5) Loop E (two for-loops outside, A inside)
############################
counter_E = 0

start = time.time()
for i in range(n):              # n
    for j in range(n):          # n
        x = n
        while x > 0:            # log n
            counter_E += 1
            x //= 2
end = time.time()

print("Loop E counter:", counter_E)
print("Loop E time:   ", end - start)
# Two nested n-loops: n^2
# Inner loop: log n
# Total: Θ(n^2 log n)
print()


############################
# 6) Loop F (three for-loops)
############################
counter_F = 0

start = time.time()
for i in range(n):              # n
    for j in range(n):          # n
        for k in range(n):      # n
            counter_F += 1
end = time.time()

print("Loop F counter:", counter_F)
print("Loop F time:   ", end - start)
# Three nested loops each running n times → Θ(n^3).
print()
```
